# Sentiment Analysis with Hugging Face Transformers

## Purpose
This notebook demonstrates how to perform sentiment analysis using various pre-trained models from the Hugging Face Transformers library. We will explore:
1.  A general-purpose sentiment analysis model.
2.  A model specifically fine-tuned for financial news sentiment analysis.
3.  A multilingual sentiment analysis model.

## 2. General-Purpose Sentiment Analysis
We start with a default sentiment analysis pipeline provided by Hugging Face, which uses the `distilbert-base-uncased-finetuned-sst-2-english` model. This model classifies text as `POSITIVE` or `NEGATIVE`.

```python
from transformers import pipeline
```

```python
model = pipeline("sentiment-analysis")
```

```python
r1 = model("i am happy with this course")
r1[0]['label']
```

```python
r2 = model("i am happy to waiting for the final certificate for showcase in my resume")
r2[0]['label']
```

## 3. Financial News Sentiment Analysis
Next, we'll use a specialized model (`mrm8488/distilroberta-finetuned-financial-news-sentiment-analysis`) that is trained to identify sentiment specifically in financial news texts. This model provides more nuanced sentiment labels relevant to financial contexts.

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification, pipeline
```

```python
tokenizer = AutoTokenizer.from_pretrained("mrm8488/distilroberta-finetuned-financial-news-sentiment-analysis")
```

```python
model = AutoModelForSequenceClassification.from_pretrained("mrm8488/distilroberta-finetuned-financial-news-sentiment-analysis")
```

```python
financial_analysis = pipeline('sentiment-analysis',
                              model=model,
                              tokenizer=tokenizer)
```

```python
t1 = financial_analysis("i am happy with this course")
t1[0]['label']
```

```python
t2 = financial_analysis('i am going on a trip')
t2[0]['label']
```

## 4. Multilingual Sentiment Analysis
Finally, we explore a multilingual sentiment analysis model (`nlptown/bert-base-multilingual-uncased-sentiment`). This model is capable of analyzing sentiment in multiple languages and often outputs sentiment as a star rating (e.g., '1 star' to '5 stars').

```python
from transformers import AutoTokenizer, AutoModelForSequenceClassification, pipeline
```

```python
tokenizer = AutoTokenizer.from_pretrained("nlptown/bert-base-multilingual-uncased-sentiment")
```

```python
nlp = AutoModelForSequenceClassification.from_pretrained('nlptown/bert-base-multilingual-uncased-sentiment')
```

```python
nlp = pipeline('sentiment-analysis',
               model=nlp,
               tokenizer=tokenizer)
```

```python
t1 = nlp("i am happy with this course")
t1[0]['label']
```

```python
t2 = nlp('i am going on a trip')
t2[0]['label']
```

```python
t3 = nlp('i am feeling very upset for my degree')
t3[0]['label']
```
