## 2. Text Generation with GPT-2

### Explanation: GPT-2 Text Generation Pipeline

We utilize the `pipeline` function from the `transformers` library for convenient text generation. We load the 'gpt2' model, which is a pre-trained language model capable of generating coherent text based on a given prompt.

`max_length` controls the maximum length of the generated sequence, and `num_return_sequences` specifies how many different sequences to generate.

```python
from transformers import pipeline

# Initialize the text generation pipeline with the GPT-2 model
generator = pipeline('text-generation', model='gpt2')
```

```python
# Define a prompt for text generation
prompt = 'once upon a time in a forest'

# Generate text based on the prompt
response = generator(prompt, max_length=50, num_return_sequences=1)

# Print the generated text
print(response[0]['generated_text'])
```

## 3. English to French Translation with MarianMT

### Explanation: MarianMT Translation Model

For English to French translation, we use the `Helsinki-NLP/opus-mt-en-fr` model. This is a MarianMT model specifically trained for this language pair. We load its `AutoTokenizer` and `AutoModelForSeq2SeqLM` separately.

Since a direct 'translation' task pipeline might not be consistently recognized across `transformers` versions, we define a custom `translate_en_to_fr` function. This function handles:
1.  **Tokenization**: Converting the input text into a format the model understands.
2.  **Generation**: Using the model to generate the translated output.
3.  **Decoding**: Converting the model's output back into human-readable text, skipping special tokens.

```python
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM

# Load the tokenizer and model for English to French translation
tokenizer_mt = AutoTokenizer.from_pretrained("Helsinki-NLP/opus-mt-en-fr")
model_mt = AutoModelForSeq2SeqLM.from_pretrained("Helsinki-NLP/opus-mt-en-fr", device_map="auto")

def translate_en_to_fr(text):
    inputs = tokenizer_mt(text, return_tensors='pt')
    # Move inputs to the same device as the model if device_map='auto' was used
    inputs = {k: v.to(model_mt.device) for k, v in inputs.items()}
    outputs = model_mt.generate(**inputs)
    return tokenizer_mt.decode(outputs[0], skip_special_tokens=True)
```

```python
# Example 1: Translate a simple sentence
english_text_1 = 'I am happy'
french_text_1 = translate_en_to_fr(english_text_1)
print(f"English: {english_text_1}")
print(f"French: {french_text_1}")
```

```python
# Example 2: Translate another sentence
english_text_2 = 'I am enjoying my day'
french_text_2 = translate_en_to_fr(english_text_2)
print(f"English: {english_text_2}")
print(f"French: {french_text_2}")
```
